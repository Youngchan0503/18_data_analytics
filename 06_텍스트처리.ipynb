```python
import pandas as pd
import numpy as np

ser = pd.Series(
    ["Apple_사과", "Banana_바나나", "Cherry_체리", np.nan],
    index=["First ", " Second", " Third", "Fourth"]
)
print(ser)
print()

ser2 = pd.Series(
    ["Apple_사과", "Banana_바나나", "Cherry_체리", np.nan],
    index=["First ", " Second", " Third", "Fourth"], dtype="string"
)
print(ser2)
```

```python
print(ser.str.lower())
print()

print(ser.str.upper())
print()

print(ser.str.len())
```

```python
print(ser.str.split("_"))
print()

print(ser.str.split("_", expand=True))
print()

print(ser.str.split("_").str.get(1))
```

```python
# 문자열 행 인덱스 추출
idx = ser.index
print(idx)

print(idx.str.strip())
print(idx.str.lstrip())
print(idx.str.rstrip())
```

```python
ser.str.replace("_",":", regex=False) # 안하면 자동으로 True
```

```python
ser.str.replace("[^a-zA-Z\\s]", "", regex=True)  # \s 는 지금 버전에서 유효하지 않다는데 확인해봐야 할 듯
```
