```python
import requests
from dotenv import load_dotenv
import os
import pandas as pd

my_key = os.environ.get("DAPADA_API_KEY")

def get_financials(stockCode, indicatorName, apiKey, consolidated=True, ttm=True):
    if consolidated:
        if ttm:
            url = f"https://api.dapada.io/company/getConsolidatedFinancialDataByTTM?indicatorName={indicatorName}&stockCode={stockCode}"
        else:
            url = f"https://api.dapada.io/company/getConsolidatedFinancialDataByCUR?indicatorName={indicatorName}&stockCode={stockCode}"


    else:
        if ttm:
            url = f"https://api.dapada.io/company/getFinancialDataByTTM?indicatorName={indicatorName}&stockCode={stockCode}"
        else:
            url = f"https://api.dapada.io/company/getFinancialDataByCUR?indicatorName={indicatorName}&stockCode={stockCode}"    


    headers = {
        "Authorization": f"{apiKey}"
    }
    response = requests.get(url, headers=headers)
    result = response.json()
    return pd.DataFrame(result)

df_ttm_consolidated = get_financials("005930", "매출액", my_key, consolidated=True, ttm=True)
df_ttm_consolidated
```

```python
import requests
from bs4 import BeautifulSoup

url = "http://apis.data.go.kr/1613000/RTMSDataSvcAptTradeDev/getRTMSDataSvcAptTradeDev"

params = {
    "serviceKey": SERVICE_KEY, 
    "LAWD_CD": "11110",
    "DEAL_YMD": 202609
}

response = requests.get(url, params=params)
print(response.status_code)
soup = BeautifulSoup(response.text, "xml")
items = soup.find_all("item")

for item in items:
    aptNm = item.find("aptNm").text
    dealAmount = item.find("dealAmount").text
    print(aptNm, dealAmount)
```
