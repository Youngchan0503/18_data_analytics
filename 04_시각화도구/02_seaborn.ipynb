```python
import seaborn as sns
titanic = sns.load_dataset("titanic")
titanic.head()
```

```python
titanic.info()
```

```python
# 회귀선이 있는 산점도
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("darkgrid")

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

sns.regplot(x="age",
            y="fare",
            data=titanic,
            ax=axes[0])

sns.regplot(x="age",
            y="fare",
            data=titanic,
            ax=axes[1],
            fit_reg=False)

plt.show()
```

```python
# 히스토그램과 커널 밀도 함수
fig, axes = plt.subplot_mosaic(
    [["top_left", "top_right"],
     ["middle_left", "middle_right"],
     ["bottom", "bottom"]],
    figsize=(15, 6),
    constrained_layout=True
)

sns.histplot(x="age", data=titanic, ax=axes["top_left"])
sns.histplot(x="age", hue="survived", data=titanic, ax=axes["top_right"])
sns.histplot(x="age", hue="survived", multiple="dodge", data=titanic, ax=axes["middle_left"])
sns.histplot(x="age", hue="survived", multiple="stack", data=titanic, ax=axes["middle_right"])
sns.histplot(x="age", hue="survived", multiple="fill", bins=10, data=titanic, ax=axes["bottom"])

plt.show()

```

```python
fig, axes = plt.subplot_mosaic([["top_left", "top_center", "right"],
                    ["bottom_left", "bottom_center", "right"]],
                    figsize=(15, 6),
                    constrained_layout=True)

sns.kdeplot(x="age", data=titanic, ax=axes["top_left"])
sns.kdeplot(x="age", hue="survived", data=titanic, ax=axes["bottom_left"])
sns.kdeplot(x="age", hue="survived", fill=True, data=titanic, ax=axes["top_center"])
sns.kdeplot(x="age", hue="survived", multiple="stack", data=titanic, ax=axes["bottom_center"])
sns.kdeplot(x="age", hue="survived", multiple="fill", bw_adjust=1.0, data=titanic, ax=axes["right"])

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

titanic = sns.load_dataset("titanic")
sns.set_style("darkgrid")

table = titanic.pivot_table(index=["sex"], columns=["class"], aggfunc=["size"], observed=False) # 집계 데이터
print(table)
type(table)
```

```python
sns.heatmap(
    table,
    annot=True,
    fmt="d",
    cmap="YlGnBu",
    linewidth=.5,
    cbar=True
)

plt.show()
```

```python
# 범주형 데이터의 산점도
sns.set_style("whitegrid")

fig, axes = plt.subplots(
    1, 2, figsize=(15, 5)
)

sns.stripplot(x="class", y="age", data=titanic, ax=axes[0])
sns.swarmplot(x="class", y="age", data=titanic, ax=axes[1], hue="class", size=4)
```

```python
# 막대그래프
sns.set_style("ticks")

fig, axes = plt.subplots(2, 2, figsize=(10, 6), constrained_layout=True)

sns.barplot(
    x="sex", y="survived", data=titanic, ax=axes[0,0]
) # 검은색 세로선은 오차범위?

sns.barplot(
    x="sex", y="survived", data=titanic, ax=axes[0,1], errorbar=None
)

sns.barplot(
    x="sex", y="survived", data=titanic, hue="class", ax=axes[1,0], errorbar=("ci", 95), estimator="median"
)

sns.barplot(
    x="sex", y="survived", data=titanic, hue="class", dodge=False, ax=axes[1,1]
)
```

```python
# 빈도그래프
sns.set_style("whitegrid")

fig, axes = plt.subplots(1, 4, figsize=(15, 5))

sns.countplot(
    x="class",
    data=titanic, ax=axes[0]
)

sns.countplot(
    x="class",
    hue="who",
    data=titanic,
    palette="Set1",
    dodge=True,
    ax=axes[1]
)

sns.countplot(
    x="class",
    hue="who",
    data=titanic,
    palette="Set2",
    dodge=False,
    ax=axes[2]
)

sns.countplot(
    x="class",
    hue="class",
    data=titanic,
    palette="Set3",
    ax=axes[3]
)
```

```python
# 박스플롯

fig, axes = plt.subplots(2, 2, figsize=(15, 10))

sns.boxplot(
    x="alive",
    y="age",
    data=titanic,
    ax=axes[0,0]
)

sns.boxplot(
    x="alive",
    y="age",
    hue="sex",
    data=titanic,
    ax=axes[0,1]
)

sns.violinplot(
    x="alive",
    y="age",
    data=titanic,
    ax=axes[1,0]
)

sns.violinplot(
    x="alive",
    y="age",
    hue="sex",
    data=titanic,
    ax=axes[1,1]
)
```

```python
# 조인트 그래프

sns.jointplot(
    x="fare",
    y="age",
    data = titanic,
)

sns.jointplot(
    x="fare",
    y="age",
    kind="reg",
    data = titanic,
)

sns.jointplot(
    x="fare",
    y="age",
    kind="hex",
    data = titanic,
)

sns.jointplot(
    x="fare",
    y="age",
    kind="kde",
    data = titanic,
)
```
