```python
import pandas as pd
import seaborn as sns

titanic = sns.load_dataset("titanic")
df = titanic.loc[0:9,["age", "fare"]]
```

```python
# 불린 인덱싱
df["age"] < 20
```

```python
cond = df["age"] < 20
df[cond]
```

```python
df.loc[cond]
```

```python
df[~cond]
```

```python
# 나이가 10대인 승객
cond = (10 <= df["age"]) & (df["age"]< 20)
df[cond]
```

```python
cond = (titanic["age"] >= 10) & (titanic["age"] < 20)
titanic[cond]
```

```python
# 나이가 10세 미만이고 여성인 승객
cond = (titanic["age"] < 10) & (titanic["sex"] == "female")
titanic[cond].loc[:,["age", "sex"]]
```

```python
# 나이가 10세 미만 또는 60세 이상인 승객
cond = (titanic["age"] < 10) | (titanic["age"] >= 60)
titanic[["age", "sex", "alone"]][cond]
```

```python
titanic[cond]
```

```python
df_under10 = titanic.loc[cond, ["age", "sex", "alone"]]
df_under10
```

```python
df["age_under_20"] = df["age"] < 20
df
```

```python
# query() 메서드 활용
# DataFrame객체.query(조건 문자열)
df = titanic.loc[0:9, ["age", "fare"]]
df
```

```python
df.query("age < 20")
```

```python
df.query("age >= 20 and fare < 15")
```

```python
# 나이가 20세 이상 또는 요금이 15 미만
df.query("age >= 20 or fare < 15")
```

```python
mask1 = titanic["embark_town"] == "Southampton"
mask2 = titanic["embark_town"] == "Queenstown"

df_boolean = titanic[mask1 | mask2]
df_boolean.head(5)
```
