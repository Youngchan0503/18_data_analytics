```python
# 시리즈 원소 선택
import pandas as pd

tup_data = ("영인", "2010-05-01", "여", True)
sr = pd.Series(tup_data, index=["이름", "생년월일", "성별", "학생여부"])
print(sr)
```

```python
# 시리즈 원소 선택
print(sr["이름"])
print(sr["성별"])

# print(sr[0]) 지금 버전에서는 지원 안함
```

```python
# print(sr[[1, 2]]) 지금 버전에서는 지원 안함
print(sr[["생년월일", "성별"]])
```

```python
print(sr)
print("================")
print(sr[1: 2]) # 되긴하는데 다른거랑 헷갈리니까 되도록 지양
```

```python
print(sr["생년월일": "학생여부"])
```

```python
print(sr[["생년월일", "학생여부"]])
```

```python
print(sr.iloc[1:3]) # 리스트 슬라이싱 처럼 마지막은 안가져옴
```
