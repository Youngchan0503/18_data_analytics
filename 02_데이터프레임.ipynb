```python
import pandas as pd

data = [
    [15, "남", "덕영중"],
    [17, "여", "수리중"],    
]

df = pd.DataFrame(data, index=["준서", "예은"], columns=["나이", "성별", "학교"])  # columns 없으면 인데스로 표시
print(df)
```

```python
df = df.rename(columns={"나이":"연령", "성별":"남녀", "학교":"소속"}) # 재할당 안하면 원본은 안바뀜
```

```python
df
```

```python
df = df.rename(index={"준서":"학생1", "예은":"학생2"})
df
```

```python
# 행/열 삭제
# 행삭제 - DataFrame객체.drop(행 인덱스 또는 배열, axis=0 또는 axis="index") # 기본적으로 axis는 행 삭제로 되어있음
# 열삭제 - DataFrame객체.drop(열 이름 또는 배열, axis=1 또는 axis="columns") # 행은 이름만으로도 찾기 가능
import pandas as pd

exam_data = {
    "수학": [90, 80, 70],
    "영어": [98, 89, 75],
    "음악": [85, 95, 100],
    "체육": [100, 90, 90]
}

df = pd.DataFrame(exam_data, index=["서준", "우현", "인아"])
df
```

```python
df2 = df.copy()
df2 = df2.drop("우현") # 재할당 안하면 원본 안바뀜
df2
```

```python
df3 = df.copy()
df3 = df3.drop(["우현", "인아"], axis=0) # 재할당 안하면 원본 안바뀜
df3
```

```python
df4 = df.copy()
df4 = df4.drop(["우현", "인아"], axis="index") # 재할당 안하면 원본 안바뀜
df4
```

```python
df5 = df.copy()
df5 = df5.drop(index = ["우현"]) # 재할당 안하면 원본 안바뀜
df5
```

```python
# 열 삭제
df2 = df.copy()
df2 = df2.drop("수학", axis=1)
df2
```

```python
df3 = df.copy()
df3 = df3.drop(["영어", "음악"], axis=1)
df3
```

```python
df4 = df.copy()
df4 = df4.drop(["영어", "음악"], axis="columns")
df4
```

```python
df5 = df.copy()
df5 = df5.drop(columns=["수학"])
df5
```
